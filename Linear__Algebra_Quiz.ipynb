## 6.) Whether the operation succeeds.

a.) (5,3)+(3,) \
  The operation will succeed according to the 1st rule of broadcasting, if both dimensions will have different dimensions, numpy will make equal dimensions by padded 1 in the shape of lesser dimension array on the left side, until both arrays have same dimensions.
  So, (3,) will become (1,3).  
  (5,3)+(1,3) but still the size of both the arrays mismatch. So, according to second law of bradcasting. If the size of both the arrays are different then the array equal to size 1 will increase by size of other dimension to match.  
  Now, (5,3)+(1,3) will become (5,3)+(5,3).   
  Therefore, broadcast works (5,3)+(3,) will SUCCEED as (5,3).  

b.) (5,3) + (5,) \
  According to the broadcast rule- size 5 and size 3 mismatch, the dimension should be equal or it should have atleast 1 in one of them.  
  So, (5,3) + (5,) will FAIL

c.) (4,1) @ (1,4) \
  In this broadcast second rule will be applied. The array with size 1 will be increase by size 4. Then the array will become  (4,4) @ (4,4).  
  So, (4,1) @ (1,4) will SUCCEED as (4,4)
